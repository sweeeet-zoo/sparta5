## 01) 뷰티 > 전체 랭킹

```python
# 1. 브라우저 열기
driver = webdriver.Chrome()
```

```python
driver.get("https://www.oliveyoung.co.kr/store/main/getBestList.do?dispCatNo=900000100100001&fltDispCatNo=&pageIdx=1&rowsPerPage=8&t_page=%EB%9E%AD%ED%82%B9&t_click=%ED%8C%90%EB%A7%A4%EB%9E%AD%ED%82%B9_%EC%A0%84%EC%B2%B4")
time.sleep(3)  # 페이지 로드 대기
```

```python
# 2. 페이지 스크롤
prev_height = driver.execute_script("return document.body.scrollHeight")
```

```python
# 1위 - 100위
while True:
    # 스크롤 내리기
    driver.execute_script("window.scrollTo(0, document.body.scrollHeight);")
    time.sleep(2)

    # 새로운 높이 가져오기
    new_height = driver.execute_script("return document.body.scrollHeight")
    if new_height == prev_height:
        break
    prev_height = new_height
```

```python
# 3. 데이터 수집
rankings = driver.find_elements(By.XPATH, '//*[@id="Container"]/div[2]/div[2]/ul/li/div/a/span')
titles = driver.find_elements(By.XPATH, '//*[@id="Container"]/div[2]/div[2]/ul/li/div/div/a/span')
products = driver.find_elements(By.XPATH, '//*[@id="Container"]/div[2]/div[2]/ul/li/div/div/a/p')
or_prices = driver.find_elements(By.XPATH, '//*[@id="Container"]/div[2]/div[2]/ul/li/div/p[1]/span[1]/span')
dis_prices = driver.find_elements(By.XPATH, '//*[@id="Container"]/div[2]/div[2]/ul/li/div/p[1]/span[2]/span')
```

```python
# 4. 텍스트 추출
print(len(rankings))
print(len(titles))
print(len(products))
print(len(or_prices))
print(len(dis_prices))
```

```python
ranking_list = [r.text for r in rankings]
ranking_list[0:5]
```

```python
title_list = [t.text for t in titles]
title_list[0:5]
```

```python
product_list = [p.text for p in products]
product_list[0:5]
```

```python
or_price_list = [op.text for op in or_prices]
or_price_list[0:5]
```

```python
dis_price_list = [dis.text for dis in dis_prices]
dis_price_list[0:5]
```

```python
# 5. 데이터 저장
olive_dic = {"랭킹": ranking_list, "브랜드": title_list, "상품명": product_list, "정가": or_price_list, "할인가": dis_price_list}
```

```python
olive_df = pd.DataFrame.from_dict(olive_dic, orient='index')
olive_df = olive_df.transpose()
olive_df.head()
```

```python
# null 값 확인
olive_df.isnull().sum()
```

```python
# 원본 데이터 저장
olive_df.to_csv("oliveyoung_ranking.csv", index=False, encoding="utf-8-sig")
```

```python
# 6. 파일 저장 및 자동화
import datetime

now = datetime.datetime.now()
file_name = f"{now.strftime('%Y-%m-%d')}_olive_output.csv"
olive_df.to_csv(file_name, index=False, encoding="utf-8-sig")
```

```python
from bs4 import BeautifulSoup
import re
```

```python
# 데이터 수집 시 다음 게시물로 넘어가기

def move_next(driver):
    right = driver.find_elements(By.CSS_SELECTOR, "div._aaqg._aaqh")
    
    if len(right) > 0:
        right[0].click()
        time.sleep(3)
        
    else:
        print("FAIL")
```

```python
# 데이터 수집 시 다음 게시물로 넘어가기 - Chat GPT 보강

def move_next_2(driver):
    try:
        right = driver.find_element(By.CSS_SELECTOR, "div._aaqg._aaqh button")
        right.click()
        time.sleep(3)

    except Exception as e:
        print("FAIL", e)
```

```python
# 데이터 수집하기 : 아이디, 본문 내용, 해시태그, 작성일자, 좋아요 수, 이미지 URL, 위치정보
def get_content(driver):
    html = driver.page_source
    soup = BeautifulSoup(html, 'html.parser')
    
    # 아이디
    ids = soup.select('div._aa1t')[0].text
    
    # 본문내용
    try:
      content = soup.select("div._a9zs")[0].text
    except:
      content = ''  
      
    # 해시태그
    tags = re.findall(r'#[^\s#,\\]+', content)
    
    # 작성일자
    date = soup.select('time._aaqe')[0]['datetime'][:10]
    
    # 좋아요
    try:
      like = soup.select('div._aacl._aaco._aacw._aacx._aada._aade')[0].findAll('span')[-1].text
    except:
      like = 0
      
    # 이미지 url
    try:
      img_urls = soup.select('div>div>div>div._aagv>img')[0]['src']
    except:
      img_urls = ''
      
    # 위치정보
    try:
      place = soup.select('div._aaqm')[0].text
    except:
      place = ''
    
    data = [ids, content, date, like, tags, img_urls, place] 
    return data
```

```python
# 데이터 수집 함수 - Chat GPT 보강

def get_content_2(driver):
    html = driver.page_source
    soup = BeautifulSoup(html, 'html.parser')
    
    # 아이디 (예외 처리 추가)
    ids = soup.select_one('a.x1i10hfl')
    ids = ids.text.strip() if ids else 'N/A'  # 값이 없으면 'N/A'로 설정
    
    # 본문내용 (예외 처리 추가)
    content = soup.select_one("div._a9zs")
    content = content.text if content else ''
    
    # 해시태그 추출
    tags = re.findall(r'#[^\s#,\\]+', content)
    
    # 작성일자 (예외 처리 추가)
    date = soup.select_one('time._aaqe')
    date = date['datetime'][:10] if date else '0000-00-00'
    
    # 좋아요 수 (예외 처리 추가)
    like_elem = soup.select_one('div._aacl._aaco._aacw._aacx._aada._aade')
    if like_elem:
        like_span = like_elem.find_all('span')
        like = like_span[-1].text if like_span else '0'
    else:
        like = '0'
    
    # 이미지 URL (예외 처리 추가)
    img_elem = soup.select_one('div > div > div > div._aagv > img')
    img_urls = img_elem['src'] if img_elem else ''
    
    # 위치정보 (예외 처리 추가)
    place = soup.select_one('div._aaqm')
    place = place.text if place else ''
    
    # 데이터 반환
    data = [ids, content, date, like, tags, img_urls, place]
    return data
```
