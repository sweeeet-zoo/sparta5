###  1.API 수집하기 - 전기자동차 충전소 정보

```python
from google.colab import userdata
#userdata.get("electric_api") #key 저장확인
```

```python
import requests
import urllib.parse

url = 'http://apis.data.go.kr/B552584/EvCharger/getChargerInfo'

# 그대로 api를 넣으면 안되고 확인해본 결과 decode를 하여 넣야아 정상작동 확인
decode_key = urllib.parse.unquote(userdata.get("electric_api"))
params ={'serviceKey' : decode_key, 'pageNo' : '1', 'numOfRows' : '10','dataType':'XML'}
response = requests.get(url, params=params)
display(response.text)
```

```python
from bs4 import BeautifulSoup

# XML 파싱
soup = BeautifulSoup(response.text, "xml")

# Pretty XML 출력
pretty_xml = soup.prettify()
print(pretty_xml)
```

```python
import xml.etree.ElementTree as ET
import pandas as pd
root = ET.fromstring(response.text)

# 모든 충전소 정보를 리스트에 저장
data = []
for item in root.findall(".//item"):
    row = {
        "statNm": item.find("statNm").text,
        "statId": item.find("statId").text,
        "chgerId": item.find("chgerId").text,
        "chgerType": item.find("chgerType").text,
        "addr": item.find("addr").text,
        "lat": item.find("lat").text,
        "lng": item.find("lng").text,
        "useTime": item.find("useTime").text,
        "busiNm": item.find("busiNm").text,
        "busiCall": item.find("busiCall").text,
        "stat": item.find("stat").text,
        "output": item.find("output").text,
        "method": item.find("method").text,
        "zcode": item.find("zcode").text,
        "parkingFree": item.find("parkingFree").text,
        "year": item.find("year").text
    }
    data.append(row)
pd.DataFrame(data)
```

### 2. 웹크롤링
- Selenium을 통한 구글 검색 자동화
- 하기 코드는 Chrome 웹브라우저를 필요로 하기 때문에, Google Colab에서는 할순 있지만 실제로 크롤링 하는 과정을 눈으로 점검하기 어려움. 따라서 vscode 권장

```python
#!pip install selenium

import selenium
print(selenium.__version__)
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.chrome.service import Service
from webdriver_manager.chrome import ChromeDriverManager

# 옵션 설정
# 하기 옵션을 넣지않으면 봇으로 감지해서 reCAPCHA와 같은 단계가 생기기 때문에, 옵션에서 사람인것 처럼 설정
options = Options()
options.add_argument("user-agent=Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/91.0.4472.124 Safari/537.36")
options.add_argument("--disable-blink-features=AutomationControlled")

# Chrome 웹 드라이버 생성
driver = webdriver.Chrome(service=Service(ChromeDriverManager().install()), options=options)


# Google 검색 페이지 열기
driver.get("https://www.google.com")

# 검색 입력 필드에 키워드 입력
search_box = driver.find_element(By.NAME, "q")
search_box.send_keys("Selenium WebDriver")
search_box.submit()
```
